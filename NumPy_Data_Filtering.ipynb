```python
import numpy as np

marks = np.array([35, 45, 50, 60, 75, 80, 90, 25, 55, 65])

print("Student marks:", marks)
print("Marks above 50:", marks[marks > 50])
```

```python
# Filtering using different conditions

above_50 = marks[marks > 50]
below_50 = marks[marks < 50]
equal_50 = marks[marks == 50]

print("Marks above 50:", above_50)
print("Marks below 50:", below_50)
print("Marks equal to 50:", equal_50)
```

```python
print("Marks above 70:", marks[marks > 70])
print("Marks below 40:", marks[marks < 40])
print("Marks greater than or equal to 60:", marks[marks >= 60])
```

```python
# Daily sales values
sales = np.array([1200, 2500, 1800, 3200, 4500, 900, 2800, 5000])

print("Sales:", sales)

# Filter sales above 2500
high_sales = sales[sales > 2500]
print("Sales above 2500:", high_sales)

# Filter sales below 2000
low_sales = sales[sales < 2000]
print("Sales below 2000:", low_sales)
```
